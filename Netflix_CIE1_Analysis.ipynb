## 6. Data Cleaning and Transformation

Data cleaning and transformation are important steps in preparing a dataset for analysis. In this section, we remove duplicate records, eliminate unnecessary identifier columns, convert data types where required and create new columns for further analysis.

The Netflix dataset stores duration as text (for example, `90 min` or `2 Seasons`). We separate the numeric duration value from its unit so that duration can be analyzed numerically. Movie minutes and TV-show seasons are different units, so they are analyzed separately where appropriate.

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Check duplicate records
print("Duplicate Records:", df_clean.duplicated().sum())

# Remove duplicate records
df_clean.drop_duplicates(inplace=True)

# Drop unnecessary identifier column
if 'show_id' in df_clean.columns:
    df_clean.drop(columns=['show_id'], inplace=True)

print("\\nDataset Shape After Cleaning:", df_clean.shape)

# Fill missing categorical/text values with Unknown
categorical_columns = df_clean.select_dtypes(include=['object', 'string']).columns
for col in categorical_columns:
    if col not in ['date_added', 'duration']:
        df_clean[col] = df_clean[col].fillna('Unknown')
```

```python
# Convert date_added to datetime and create year/month columns
df_clean['date_added'] = pd.to_datetime(
    df_clean['date_added'].astype('string').str.strip(),
    errors='coerce'
)

df_clean['date_added_year'] = df_clean['date_added'].dt.year
df_clean['date_added_month'] = df_clean['date_added'].dt.month

# Extract numeric duration and duration unit
duration_parts = df_clean['duration'].astype('string').str.extract(
    r'(?P<duration_value>\\d+)\\s*(?P<duration_unit>.*)'
)

df_clean['duration_value'] = pd.to_numeric(
    duration_parts['duration_value'], errors='coerce'
)
df_clean['duration_value'] = df_clean['duration_value'].astype(float)
df_clean['duration_unit'] = duration_parts['duration_unit'].str.strip()

# Normalize duration unit labels
df_clean['duration_unit'] = df_clean['duration_unit'].replace({
    'Season': 'Seasons',
    'min': 'Minutes'
})

# Convert release_year to numeric
df_clean['release_year'] = pd.to_numeric(
    df_clean['release_year'], errors='coerce'
)

# Display updated dataset
display(df_clean.head())
```

```python
# Display data types after transformation
print(df_clean.dtypes)
```

## 7. Handling Missing Values Using Mean Imputation

Mean imputation is a data-cleaning technique used to replace missing numerical values with the mean of their respective columns.

In this Netflix dataset, `duration_value` is the numerical duration extracted from the original duration field. Movie durations are measured in minutes, while TV-show durations are measured in seasons. Therefore, missing duration values are imputed using the mean for the corresponding content type, rather than mixing minutes and seasons.

Date fields and categorical labels are not suitable for numerical mean imputation. Missing `date_added` values are retained as unknown dates.

```python
# Identify numerical columns
numerical_columns = df_clean.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numerical_columns.tolist())
```

```python
# Calculate missing values before imputation
print("Missing values before imputation:")
print(df_clean[numerical_columns].isnull().sum())

# Mean imputation for duration within each content type
for content_type in df_clean['type'].dropna().unique():
    mask = (
        (df_clean['type'] == content_type) &
        (df_clean['duration_value'].isnull())
    )
    group_mean = df_clean.loc[
        df_clean['type'] == content_type, 'duration_value'
    ].mean()
    if pd.notna(group_mean):
        df_clean.loc[mask, 'duration_value'] = group_mean

# Mean imputation for any remaining suitable numeric columns
for col in numerical_columns:
    if col != 'duration_value' and df_clean[col].isnull().any():
        mean_value = df_clean[col].mean()
        df_clean[col] = df_clean[col].fillna(mean_value)

print("\\nMissing values after numerical mean imputation:")
print(df_clean[numerical_columns].isnull().sum())
```

```python
# Check remaining missing values in the dataset
print("Total Missing Values:", df_clean.isnull().sum().sum())
print("\\nRemaining missing values by column:")
display(df_clean.isnull().sum().to_frame(name="Remaining Missing Values"))
```

## 8. Exploratory Data Analysis (EDA)

Exploratory Data Analysis is the process of examining and analyzing a dataset to discover patterns, relationships and trends. In this section, we explore Netflix content types, release years, genres, ratings and durations.

Since the Netflix dataset does not contain a popularity score, the analysis uses measurable Netflix fields such as content counts and release years instead of Spotify popularity.

```python
# Calculate average values of numerical attributes

display(df_clean.describe().T)
```

```python
# Display the 10 most recent release-year titles

recent_titles = df_clean.sort_values(
    by='release_year',
    ascending=False
).head(10)

display(recent_titles[['title', 'type', 'release_year', 'rating']])
```

```python
# Calculate the number of titles in each genre
# A title can belong to more than one genre/category.

genre_counts = (
    df_clean['listed_in']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
)

display(genre_counts.head(10))
```

## 9. Data Visualization

Data visualization helps represent complex information through graphical formats. In this section, we use Matplotlib and Seaborn to visualize Netflix content distribution, genres, release years and durations.

Each chart uses fields available in the Netflix dataset. Movie duration is plotted separately from TV-show season counts because the two units are not directly comparable.

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df_clean, x='type', color='steelblue')

plt.title('Distribution of Netflix Content Types')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')

plt.show()
```

```python
plt.figure(figsize=(12, 6))

top_genres = genre_counts.head(10).sort_values(ascending=True)

sns.barplot(
    x=top_genres.values,
    y=top_genres.index,
    color='steelblue'
)

plt.title('Top 10 Most Common Netflix Genres')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')

plt.show()
```

```python
# Scatter plot: release year vs movie duration in minutes
movie_data = df_clean[
    (df_clean['type'] == 'Movie') &
    (df_clean['duration_unit'] == 'Minutes') &
    (df_clean['duration_value'].notna())
]

plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=movie_data,
    x='release_year',
    y='duration_value',
    alpha=0.4,
    color='steelblue'
)

plt.title('Movie Release Year vs Movie Duration')
plt.xlabel('Release Year')
plt.ylabel('Duration (Minutes)')

plt.show()
```

```python
# Distribution of movie duration in minutes
plt.figure(figsize=(10, 5))

sns.histplot(
    data=movie_data,
    x='duration_value',
    bins=30,
    kde=True,
    color='steelblue'
)

plt.title('Distribution of Netflix Movie Duration')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Number of Movies')

plt.show()
```

```python
# Select numerical columns
numeric_df = df_clean.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Between Netflix Numerical Attributes')

plt.show()
```
